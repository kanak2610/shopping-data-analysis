```python
import pandas as pd
```

```python
df = pd.read_csv("data/shopping.csv")
```

```python
# Show first 5 rows
print(df.head())
```

```python
# Show last 5 rows
print(df.tail())
```

```python
# Display total rows and columns
print(df.shape)
```

```python
# Show column names
print(df.columns)
```

```python
# Show datatype of each column
print(df.dtypes)
```

```python
# Display complete dataset information
df.info()
```

```python
# Count missing values in each column
print(df.isnull().sum())
```

```python
# Select numeric columns
num_cols = df.select_dtypes(include=['number']).columns

# Display numeric columns
print(num_cols)
```

```python
# Replace missing numeric values with 0
df[num_cols] = df[num_cols].fillna(0)
```

```python
# Select text/string columns
str_cols = df.select_dtypes(include=['object', 'string']).columns

# Display string columns
print(str_cols)
```

```python
# Replace missing text values with empty string
df[str_cols] = df[str_cols].fillna("")
```

```python
# Check missing values again
print(df.isnull().sum())
```

```python
# Select title and final_price columns
selected_columns = df[['title', 'final_price']]

# Display selected columns
print(selected_columns.head())
```

```python
# Convert final_price column to numeric
df['final_price'] = pd.to_numeric(df['final_price'], errors='coerce')

# Display datatype
print(df['final_price'].dtype)
```

```python
# Filter rows where final_price > 1000
filtered_data = df[df['final_price'] > 1000]

# Display filtered rows
print(filtered_data.head())
```

```python
# Count duplicate rows
print(df.duplicated().sum())
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)
```

```python
# Check duplicate rows again
print(df.duplicated().sum())
```

```python
# Create discount_price column
df['discount_price'] = df['final_price'] * 0.9

# Display updated dataset
print(df[['title', 'final_price', 'discount_price']].head())
```

```python
# Show statistical summary
print(df.describe())
```

```python
# Show statistical summary
print(df.describe())
```

```python
# Count unique values in each column
print(df.nunique())
```

```python
# Save cleaned dataset into CSV file
df.to_csv("cleaned_shopping.csv", index=False)

# Display success message
print("Cleaned dataset saved successfully.")
```
